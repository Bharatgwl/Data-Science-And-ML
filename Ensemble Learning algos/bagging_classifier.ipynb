```python
param_grid = {
    'estimator': [DecisionTreeClassifier(), SVC(), LogisticRegression(), KNeighborsClassifier()],
    'n_estimators': [100, 200],  # Number of base learners
    'max_samples': [0.1, 0.2, 0.4, 0.5, 0.7],  # Proportion of training data per estimator
    'max_features': [0.1, 0.2, 0.4, 0.5, 0.7],  # Proportion of features per estimator
    'bootstrap': [True, False],  # Sampling with/without replacement
    'bootstrap_features': [True, False]  # Sampling features with/without replacement
}


from sklearn.model_selection import GridSearchCV
bgclf = BaggingClassifier(random_state=42)
gd = GridSearchCV(bgclf, param_grid,cv=5,n_jobs=-1)
```

```python

gd.fit(X_train,y_train)
```
