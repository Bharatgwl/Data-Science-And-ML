```python
# how to create multiindex dataframe
# 1. pd.MultindexFromtuple
index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('cse',2023),('cse',2020),('ece',2020),('ece',2022),('ece',2021),('ece',2018),('ece',2017)]
multiindex= pd.MultiIndex.from_tuples(index_val)
multiindex.levels  # decouple the branch and year 

# 2. pd.multiindex.from_product

multiindex = pd.MultiIndex.from_product([['cse', 'ece'], [2017, 2018, 2019, 2020, 2021, 2022, 2023]])
```
