```python
import numpy as np
import matplotlib.pyplot as plt

# Toy 2D dataset
X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2.0, 1.6],
    [1.0, 1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

# Step 1: Mean centering
X_meaned = X - np.mean(X, axis=0)

# Step 2: Covariance matrix
cov_mat = np.cov(X_meaned, rowvar=False)

# Step 3: Eigen decomposition
eigen_vals, eigen_vecs = np.linalg.eigh(cov_mat)

# Step 4: Sort eigenvalues and vectors
sorted_idx = np.argsort(eigen_vals)[::-1]
eigen_vals = eigen_vals[sorted_idx]
eigen_vecs = eigen_vecs[:, sorted_idx]

# Step 5: Project data
projected = X_meaned.dot(eigen_vecs)

# Plot 1: Original Data
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], color='blue', alpha=0.6)
plt.title("Original 2D Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.savefig("original_data.png", dpi=300)
plt.close()

# Plot 2: Eigenvectors overlay
plt.figure(figsize=(6, 5))
plt.scatter(X_meaned[:, 0], X_meaned[:, 1], alpha=0.6)
origin = [0, 0]
for i in range(len(eigen_vecs)):
    vec = eigen_vecs[:, i] * np.sqrt(eigen_vals[i]) * 2
    plt.quiver(*origin, vec[0], vec[1], angles='xy', scale_units='xy', scale=1, color=['green', 'magenta'][i])
plt.title("Eigenvectors on Centered Data")
plt.xlabel("X (centered)")
plt.ylabel("Y (centered)")
plt.grid(True)
plt.axis('equal')
plt.savefig("eigenvectors_plot.png", dpi=300)
plt.close()

# Plot 3: PCA-transformed data
plt.figure(figsize=(6, 5))
plt.scatter(projected[:, 0], projected[:, 1], color='orange', alpha=0.6)
plt.title("Data in Principal Component Space")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.axis('equal')
plt.savefig("pca_transformed.png", dpi=300)
plt.close()

print("Plots saved: original_data.png, eigenvectors_plot.png, pca_transformed.png")
```
