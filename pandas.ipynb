```python
from io import StringIO, BytesIO
```

```python
data = ('col1,col2,col3\n'
       'x,y,1\n'
       'a,b,c\n'
       'c,d,3')
```

```python
type (data)
```

```python
pd.read_csv(StringIO(data)) #converting th str
```

```python
## reading the specific column using pd readcsv
df = pd.read_csv(StringIO(data), usecols=['col1','col2'])
df
```

```python
df.to_csv('sample.csv')

df
```

# Read JSON to CSV

```python
data = '{"username": "bharat_gwl", "email": "bharatgusaiwal731@gmail.com", "age": 19,"password": "1234"}'

# Normalize nested JSON if needed (e.g., createdAt and updatedAt)
df = pd.read_json(StringIO(data),orient='index')

df
```

```python
df.to_csv('jsonData.csv') # converting to csv , saves in the files
```
